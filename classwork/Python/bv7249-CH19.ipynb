```python
import numpy as np
import matplotlib.pyplot as plt

def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

def analytical_sol(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

h = 0.001          
t_end = 0.02
t_steps = np.arange(0, t_end + h, h)

y_euler = np.zeros(len(t_steps))
y_euler[0] = 0  

for i in range(0, len(t_steps) - 1):
    y_euler[i+1] = y_euler[i] + h * f(t_steps[i], y_euler[i])

t_fine = np.linspace(0, t_end, 500)
y_analytical = analytical_sol(t_fine)

plt.figure(figsize=(10, 6))
plt.plot(t_fine, y_analytical, 'r-', label='Analytical Solution', linewidth=2)
plt.plot(t_steps, y_euler, 'bo--', label="Euler's Forward", markersize=4, alpha=0.7)

plt.title("Euler's Forward vs Analytical Solution")
plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.legend()
plt.grid(True)
plt.show()
```
