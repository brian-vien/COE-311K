```python
import numpy as np

# Define the function f(t, y)
def f(t, y):
    return y**2 - 10

# Define the derivative of f with respect to y (f prime)
def fprime(y):
    return 2 * y

def backward_euler_step(f, y, t, h, tol=1e-6, max_iter=50):
    y_next = y  # Initial guess
    for _ in range(max_iter):
        # f(t + h, y_next) is the value of the ODE at the next step
        g_val = y_next - y - h * f(t + h, y_next)
        # g_prime is the derivative of the root-finding function g
        g_prime_val = 1 - h * fprime(y_next) 
        
        if abs(g_prime_val) < 1e-12:
            break
            
        y_new = y_next - g_val / g_prime_val
        
        if abs(y_new - y_next) < tol:
            return y_new 
        y_next = y_new
    return y_next

def backward_euler_adaptive(f, y0, t0, tf, h_init, tol=1e-6, max_iter=50):
    t_values = [t0]
    y_values = [y0]
    h = h_init
    t = t0
    y = y0
    
    while t < tf:
        if t + h > tf:
            h = tf - t
            
        # Single step with h
        y_full = backward_euler_step(f, y, t, h, tol, max_iter)
        
        # Two half-steps with h/2
        h_half = h / 2
        y_half_1 = backward_euler_step(f, y, t, h_half, tol, max_iter)
        y_half_2 = backward_euler_step(f, y_half_1, t + h_half, h_half, tol, max_iter)
        
        # Error estimation
        error = abs(y_full - y_half_2)
        
        # Adjust step size
        if error > tol and h > 1e-10: 
            h /= 2
            continue 
        elif error < tol / 10: 
            h *= 2
            
        t += h
        y = y_half_2
        t_values.append(t)
        y_values.append(y)
        
    return np.array(t_values), np.array(y_values)

t_res, y_res = backward_euler_adaptive(f, y0=3.0, t0=0, tf=2, h_init=0.1)

print(f"Final y at t={t_res[-1]}: {y_res[-1]}")
```
